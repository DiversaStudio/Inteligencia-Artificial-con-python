## 1. La estructura general

Casi toda línea de Python que usaremos tiene esta forma:

```
variable  =  objeto.método(argumentos)
```

| Parte | Qué es | Ejemplo |
|---|---|---|
| `variable` | dónde guardamos el resultado | `limpio` |
| `=` | "guarda en" | |
| `objeto` | el dato con el que trabajamos | un texto, una lista, una tabla |
| `.método` | la acción que hace el objeto | `.replace` |
| `(argumentos)` | los detalles que necesita la acción | `("Quito", "[CIUDAD]")` |

Leámoslo en español: *"guarda en `limpio` el resultado de que `testimonio` reemplace 'Quito' por '[CIUDAD]'"*.

````
testimonio = "Me llamo Rosa, vivo en Quito desde 2019 y trabajo en el mercado"

limpio = testimonio.replace("Rosa", "[NOMBRE]")
limpio = limpio.replace("Quito", "[CIUDAD]")

print(testimonio)
print(limpio)
````

Acabamos de **anonimizar** un testimonio con dos líneas. Es una tarea real de investigación: antes de subir entrevistas a cualquier herramienta de IA, quitamos los datos que identifican a las personas.

Otros métodos de texto con la misma estructura:

````
minusculas = testimonio.lower()          # sin argumentos, pero con paréntesis
palabras   = testimonio.split(" ")       # argumento: dónde cortar
cuenta     = testimonio.count("o")       # argumento: qué contar

print(minusculas)
print(palabras)
print(cuenta)
````

## 2. Cuatro formas que vas a ver siempre

| Forma | Cómo se escribe | Qué hace | ¿Paréntesis? |
|---|---|---|---|
| **Atributo** | `objeto.atributo` | una característica del objeto | No |
| **Método** | `objeto.método()` | una acción que hace el objeto | Sí |
| **Función** | `función(objeto)` | una herramienta de Python que recibe el objeto | Sí |
| **Selección** | `objeto[...]` | tomar una parte del objeto | Corchetes |

Veámoslas primero con una lista:

````
entrevistas = ["Rosa", "Luis", "Mariana", "Andrés"]

print(len(entrevistas))        # FUNCIÓN: len recibe la lista
print(entrevistas[0])          # SELECCIÓN: el primer elemento
entrevistas.append("Carmen")   # MÉTODO: la lista se agrega un elemento
print(entrevistas)
````

Los **atributos** aparecen sobre todo en objetos más grandes, como las tablas de pandas. Lo vemos en la siguiente sección.

## 4. Trabajar con muchos archivos

En las próximas semanas no tendremos un solo dato, sino **carpetas** llenas de fotos, PDF o audios. El patrón es siempre el mismo:

1. Listar los archivos de una carpeta
2. Recorrerlos con `for`
3. Aplicar algo a cada uno
4. Guardar los resultados en una tabla

Primero creamos una carpeta de ejemplo con algunas transcripciones:

````
import os

os.makedirs("transcripciones", exist_ok=True)

textos = {
    "E01.txt": "Llegamos a Quito buscando trabajo y nos quedamos en el sur.",
    "E02.txt": "En el barrio no hay transporte después de las ocho de la noche.",
    "E03.txt": "La comunidad organizó una minga para arreglar el camino al colegio.",
}

for nombre, contenido in textos.items():
    with open(f"transcripciones/{nombre}", "w", encoding="utf-8") as archivo:
        archivo.write(contenido)

print(os.listdir("transcripciones"))
````

Ahora el patrón completo: listar, recorrer, leer y guardar en una tabla.

````
import glob

rutas = sorted(glob.glob("transcripciones/*.txt"))   # todos los .txt de la carpeta
print(rutas)

resultados = []
for ruta in rutas:
    with open(ruta, encoding="utf-8") as archivo:
        texto = archivo.read()
    resultados.append({
        "archivo": os.path.basename(ruta),
        "palabras": len(texto.split()),
        "texto": texto,
    })

tabla = pd.DataFrame(resultados)
tabla
````

**Guarda este patrón.** En la semana de audio, la línea que lee el archivo se convertirá en `modelo.transcribe(ruta)`; en la de documentos, en `lector.readtext(ruta)`. Todo lo demás queda igual.

`*.txt` es un **comodín**: significa "cualquier nombre que termine en .txt". Para imágenes usaremos `*.jpg`, para audios `*.mp3`.

## 5. Cuando algo falla: `try` / `except`

Si procesas 100 archivos y uno está dañado, no quieres que todo se detenga. `try` intenta algo; si falla, `except` dice qué hacer en su lugar.

````
rutas_prueba = ["transcripciones/E01.txt", "transcripciones/NO_EXISTE.txt", "transcripciones/E03.txt"]

for ruta in rutas_prueba:
    try:
        with open(ruta, encoding="utf-8") as archivo:
            print("✔", ruta, "→", len(archivo.read().split()), "palabras")
    except FileNotFoundError:
        print("✘ No encontré", ruta, "- sigo con el siguiente")
````

### Errores frecuentes y qué significan
| Error | Significa | Revisa |
|---|---|---|
| `NameError` | esa variable no existe | ¿ejecutaste la celda donde se crea? ¿está bien escrita? |
| `KeyError` | esa clave o columna no existe | mayúsculas: `"Provincia"` ≠ `"provincia"` |
| `IndexError` | pediste una posición que no hay | las listas empiezan en 0 |
| `FileNotFoundError` | no encuentra el archivo | la ruta y el nombre |
| `ModuleNotFoundError` | la librería no está instalada | usa `!pip install` |
| `SyntaxError` / `IndentationError` | algo está mal escrito | comillas, paréntesis, `:` y sangría |

## 7. Leer lo que devuelve un modelo de IA

Los modelos devuelven sus resultados como **diccionarios y listas, uno dentro de otro**. Este es un resultado simulado con la misma forma que el de Whisper, el modelo de transcripción que usaremos en la semana 4:

````
resultado = {
    "text": "Buenos días. Yo llegué a Quito en 2019. Trabajo en el mercado.",
    "language": "es",
    "segments": [
        {"start": 0.0, "end": 1.8, "text": "Buenos días."},
        {"start": 1.8, "end": 4.5, "text": "Yo llegué a Quito en 2019."},
        {"start": 4.5, "end": 6.9, "text": "Trabajo en el mercado."},
    ],
}

print(resultado["language"])          # selección en un diccionario
print(resultado["segments"][1])       # selección: lista dentro del diccionario
print(resultado["segments"][1]["text"])  # y un diccionario dentro de la lista
````

Para leer selecciones encadenadas, ve de izquierda a derecha: del resultado → toma los segmentos → toma el segundo (posición 1) → toma su texto.

````
for seg in resultado["segments"]:
    print(f"[{seg['start']:>4} - {seg['end']:>4} s] {seg['text']}")
````

````
# Convertirlo en tabla, igual que cualquier otra lista de diccionarios
pd.DataFrame(resultado["segments"])
````

Y esta es la forma típica de un modelo de detección de objetos en imágenes (como YOLO, semana 2):

````
detecciones = [
    {"objeto": "persona", "confianza": 0.94},
    {"objeto": "persona", "confianza": 0.88},
    {"objeto": "bicicleta", "confianza": 0.71},
    {"objeto": "perro", "confianza": 0.42},
]

# Quedarnos solo con lo que el modelo detectó con más de 60 % de confianza
for d in detecciones:
    if d["confianza"] > 0.6:
        print(d["objeto"], d["confianza"])
````
