### Pausa · Cómo leer una línea de Python

Fíjate que `tabla.head()` lleva paréntesis y `tabla.shape` no. No es casualidad. Casi todo lo que escribimos se arma con estas piezas:

| Pieza | Cómo se ve | Qué es | En español sería… |
|---|---|---|---|
| **Objeto** | `tabla` | el dato con el que trabajo | el sustantivo |
| **Atributo** | `tabla.shape` | una característica del objeto. **Sin paréntesis** | un adjetivo: "la tabla *tiene* 12 filas" |
| **Método** | `tabla.head()` | una acción que el objeto sabe hacer. **Con paréntesis** | un verbo: "la tabla *muestra* sus primeras filas" |
| **Argumento** | `tabla.head(3)` | el detalle que le doy a la acción, **dentro** del paréntesis | "muestra **3** filas" |
| **Función** | `len(tabla)` | una acción que no pertenece a un objeto: el objeto va dentro | "cuenta **la tabla**" |
| **Selección** | `tabla["edad"]` | tomar una parte. **Con corchetes** | "de la tabla, la columna edad" |

Regla corta: **el punto** significa "de este objeto", **los paréntesis** significan "haz algo" y **los corchetes** significan "toma una parte".

El argumento cambia lo que hace el método. Compara:

````
tabla.head()
````

````
tabla.head(3)
````

Un método puede recibir varios argumentos, separados por comas. Algunos tienen nombre:

````
tabla.sort_values("edad", ascending=False)
````

Aquí `sort_values` es el método (ordenar), `"edad"` es el primer argumento (por qué columna) y `ascending=False` es un argumento con nombre (de mayor a menor).

Ya lo venías usando: en `pd.read_csv("entrevistas.csv")`, `pd` es la librería, `read_csv` es la acción y `"entrevistas.csv"` es el argumento: qué archivo leer.

## 4. Recorrer la base con un `for`

Una columna de la tabla funciona como una lista. Con `for` puedo hacer lo mismo con cada fila:

````
for testimonio in tabla["testimonio"]:
    print(testimonio)
````

Se lee así: **para cada** `testimonio` **en** la columna `"testimonio"`, imprímelo.

- `testimonio` es el elemento que tengo "en la mano" en cada vuelta. Ese nombre lo invento yo.
- Los **dos puntos** y la **sangría** (los 4 espacios) indican qué se repite.
- Hay tantas vueltas como filas tiene la tabla: 12.

### Guardar lo que pasa en cada vuelta

Este es el patrón que vamos a usar todo el curso con fotos, documentos y audios:

1. **Antes** del `for`: una lista vacía para guardar resultados.
2. **Dentro** del `for`: hago algo con cada elemento y guardo una ficha con `append`.
3. **Después** del `for`: convierto los resultados en una tabla nueva.

````
resultados = []

for testimonio in tabla["testimonio"]:
    ficha = {"testimonio": testimonio, "palabras": len(testimonio.split())}
    resultados.append(ficha)

conteo = pd.DataFrame(resultados)
conteo
````

`testimonio.split()` corta el texto en palabras y `len(...)` las cuenta.
