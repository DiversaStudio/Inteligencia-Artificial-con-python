## 1. Imprimir y comentar
`print()` muestra algo en pantalla. Todo lo que va después de `#` es un **comentario**: Python lo ignora, sirve para explicar el código a otras personas (y a tu yo del futuro).

```
# Esto es un comentario
print("Hola, ciencias sociales")
```

## 2. Variables y tipos de datos
Una **variable** es una etiqueta que guarda un valor. Python reconoce distintos **tipos**:

| Tipo | Ejemplo | Para qué |
|---|---|---|
| `str` (texto) | `"Quito"` | nombres, testimonios, transcripciones |
| `int` (entero) | `34` | edades, conteos |
| `float` (decimal) | `12.5` | duraciones, porcentajes |
| `bool` (verdadero/falso) | `True` | ¿firmó el consentimiento? |

```
participante = "Rosa"
edad = 34
duracion_min = 42.5
consentimiento = True

print(participante, edad, duracion_min, consentimiento)
print(type(participante), type(edad), type(duracion_min), type(consentimiento))
```

Con un **f-string** (una `f` antes de las comillas) puedes mezclar texto y variables:

```
print(f"{participante} tiene {edad} años y su entrevista duró {duracion_min} minutos.")
```

Los textos también tienen sus propias herramientas:

```
testimonio = "Llegamos a Quito en 2019 buscando trabajo"

print(testimonio.lower())        # todo en minúsculas
print(len(testimonio))           # cuántos caracteres tiene
print("trabajo" in testimonio)   # ¿aparece esta palabra?
print(testimonio.split())        # separar en palabras
```

## 3. Listas
Una **lista** guarda varios valores en orden, entre corchetes `[ ]`. Python empieza a contar desde **0**.

```
entrevistas = ["Rosa", "Luis", "Mariana", "Andrés"]

print(entrevistas[0])      # el primer elemento
print(entrevistas[-1])     # el último
print(len(entrevistas))    # cuántos hay

entrevistas.append("Carmen")   # agregar uno al final
print(entrevistas)
```

## 4. Diccionarios
Un **diccionario** guarda pares **clave: valor**, entre llaves `{ }`. Es como una ficha de registro.

```
ficha = {
    "nombre": "Rosa",
    "edad": 34,
    "ciudad": "Quito",
    "testimonio": "Llegamos a Quito en 2019 buscando trabajo"
}

print(ficha["ciudad"])
ficha["idioma"] = "español"     # agregar un campo nuevo
print(ficha)
```

## 5. Tomar decisiones con `if`
`if` ejecuta un bloque solo si se cumple una condición. Fíjate en los **dos puntos** y en la **sangría** (los espacios al inicio): así Python sabe qué está dentro del `if`.

```
if ficha["edad"] >= 18:
    print("Participante adulta: puede firmar su propio consentimiento.")
else:
    print("Menor de edad: se requiere consentimiento de su representante.")
```

## 6. Repetir con `for`
`for` repite una acción para cada elemento de una lista. Es la base de todo lo que haremos con IA: procesar **muchas** imágenes, documentos o audios con el mismo código.

```
for nombre in entrevistas:
    print(f"Procesando la entrevista de {nombre}...")
```

Combinando todo: una lista de fichas, recorrida con `for`, con una decisión `if` adentro.

```
registros = [
    {"nombre": "Rosa",    "testimonio": "Llegamos a Quito en 2019 buscando trabajo"},
    {"nombre": "Luis",    "testimonio": "En mi barrio no hay transporte después de las ocho"},
    {"nombre": "Mariana", "testimonio": "Conseguí trabajo gracias a una vecina"},
]

palabra = "trabajo"
for r in registros:
    if palabra in r["testimonio"].lower():
        print(f"{r['nombre']} menciona '{palabra}'")
```

## 7. Funciones
Una **función** guarda un bloque de código con un nombre para reutilizarlo. Se define con `def` y devuelve un resultado con `return`.

```
def contar_palabras(texto):
    return len(texto.split())

for r in registros:
    print(r["nombre"], "→", contar_palabras(r["testimonio"]), "palabras")
```

## 8. Librerías
Una **librería** es código que otras personas escribieron y que podemos usar con `import`. La primera que usaremos es **pandas**, para trabajar con tablas (como una hoja de cálculo).

```
import pandas as pd

tabla = pd.DataFrame(registros)
tabla["palabras"] = tabla["testimonio"].apply(contar_palabras)
tabla
```

En las próximas semanas usaremos librerías como **Whisper** (audio), **EasyOCR** (documentos) o **YOLO** (imágenes) exactamente así: `import`, cargar un modelo y aplicarlo a nuestros datos con un `for`.

```
# 🟢 Básico
registros.append({"nombre": "Diana", "testimonio": "Trabajo en un barrio del norte de Quito"})

# 🟡 Intermedio
palabra = "barrio"
contador = 0
for r in registros:
    if palabra in r["testimonio"].lower():
        contador = contador + 1
print(f"{contador} testimonios mencionan '{palabra}'")

# 🔴 Avanzado
def buscar(registros, palabra):
    encontrados = []
    for r in registros:
        if palabra in r["testimonio"].lower():
            encontrados.append(r["nombre"])
    return encontrados

print(buscar(registros, "trabajo"))
```
